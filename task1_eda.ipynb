```python
Q1 = df["Global_Sales"].quantile(0.25)
Q3 = df["Global_Sales"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Global_Sales"] < lower_bound) |
    (df["Global_Sales"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Potential outlier rows:", len(outliers))
```

```python

sales_columns = [
    "NA_Sales",
    "EU_Sales",
    "JP_Sales",
    "Other_Sales"
]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for column, ax in zip(sales_columns, axes.flatten()):
    sns.histplot(data=df, x=column, bins=40, ax=ax)
    ax.set_title(f"Distribution of {column}")
    ax.set_xlabel("Sales (millions of units)")
    ax.set_ylabel("Number of Games")

plt.tight_layout()

plt.savefig(
    figures_dir / "regional_sales_distributions.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

**Observation:** The regional sales distributions are strongly right-skewed. In North America, Europe, Japan, and other regions, most games have relatively low sales, while a smaller number have substantially higher sales. This pattern suggests that high sales are concentrated among a limited number of games. The different horizontal axis ranges also show that the regional sales values span different ranges. Further statistical comparisons are needed to compare typical sales across regions.

```python

sales_columns = [
    "NA_Sales",
    "EU_Sales",
    "JP_Sales",
    "Other_Sales",
    "Global_Sales"
]

correlation_matrix = df[sales_columns].corr()

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Between Regional and Global Video Game Sales")

plt.tight_layout()

plt.savefig(
    figures_dir / "sales_correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

**Observation:** The correlation heatmap shows positive relationships among all regional sales variables. North American sales have the strongest correlation with global sales (0.94), followed by European sales (0.90). North American and European sales are also strongly correlated (0.77). The weakest relationship is between Japanese sales and other-region sales (0.29). Since global sales incorporates regional sales, its strong correlations with regional sales are expected and should be interpreted cautiously.

```python

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

pairs = [
    ("NA_Sales", "EU_Sales"),
    ("NA_Sales", "JP_Sales"),
    ("EU_Sales", "JP_Sales")
]

for ax, (x_col, y_col) in zip(axes, pairs):
    sns.scatterplot(
        data=df,
        x=x_col,
        y=y_col,
        alpha=0.35,
        s=18,
        ax=ax
    )
    ax.set_title(f"{x_col} vs {y_col}")
    ax.set_xlabel(f"{x_col} (millions)")
    ax.set_ylabel(f"{y_col} (millions)")

plt.tight_layout()

plt.savefig(
    figures_dir / "regional_sales_scatterplots.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

**Observation:** The scatter plots show that most video games have relatively low sales across North America, Europe, and Japan, as indicated by the dense clusters of points near the bottom-left corner. North American and European sales show a stronger positive relationship (correlation = 0.77), while the relationships between North American and Japanese sales (0.45) and European and Japanese sales (0.44) are weaker. A few games have substantially higher sales than most others, consistent with the potential outliers identified earlier. These relationships indicate association, not causation.

### Release Year Validation

The `Year` column contains no non-whole-number values and no recorded years outside the observed range of 1980–2020. Therefore, no corrections to existing year values were necessary. However, 271 release-year values are missing. These values will be retained as missing rather than replaced with guessed years.

```python
regional_cols = [
    "NA_Sales",
    "EU_Sales",
    "JP_Sales",
    "Other_Sales"
]

df["Calculated_Global_Sales"] = df[regional_cols].sum(axis=1)

df["Sales_Difference"] = (
    df["Global_Sales"] - df["Calculated_Global_Sales"]
)

print("Rows with a sales difference:", (df["Sales_Difference"].abs() > 0.01).sum())
print("Largest absolute difference:", df["Sales_Difference"].abs().max())
```

```python
print("Rows with difference > 0.01:",
      (df["Sales_Difference"].abs() > 0.01).sum())

print("\nLargest differences:")
print(
    df.loc[
        df["Sales_Difference"].abs() > 0.01,
        ["Name", "Global_Sales", "Calculated_Global_Sales", "Sales_Difference"]
    ].head(10)
)
```

```python
# Round the absolute differences to two decimal places
rounded_diff = df["Sales_Difference"].abs().round(2)

print("Rows with difference of 0.01:",
      (rounded_diff == 0.01).sum())

print("Rows with difference of 0.02:",
      (rounded_diff == 0.02).sum())

print("Rows with difference greater than 0.02:",
      (rounded_diff > 0.02).sum())
```

### Regional and Global Sales Consistency

Regional sales were summed and compared with the recorded `Global_Sales` values. The differences were small: 4,501 rows had an absolute difference of 0.01 million units, and 10 rows had a difference of 0.02 million units. No rows had a rounded absolute difference greater than 0.02 million units.

These minor discrepancies are consistent with rounding in the recorded sales figures. Therefore, the original sales values were retained without modification.

```python
df.drop(
    columns=["Calculated_Global_Sales", "Sales_Difference"],
    inplace=True
)

print("Current dataset shape:", df.shape)
print("Current columns:", df.columns.tolist())
```
